```python
import os
from google.generativeai import GenerativeModel, configure
import json
```

```python
# 2. Sample user inputs
symptoms = input("Enter patient symptoms (comma separated): ")
age = input("Enter patient age: ")
mobility = input("Describe mobility issues: ")
allergies = input("List known allergies (comma separated): ")
conditions = input("List chronic conditions (comma separated): ")
surgeries = input("Recent surgeries (if any): ")
immunization = input("Immunization status: ")
other_notes = input("Any other notes: ")
city = input("City: ")
state = input("State: ")
mode_of_care = input("Preferred mode of care (In person / Telehealth / Either): ")
```

```python
# 3. Hospital data
hospital_data = [
    { "Name": "University of Cincinnati Medical Center", "Latitude": 39.1412, "Longitude": -84.5059 },
    { "Name": "Cincinnati VA Medical Center", "Latitude": 39.1398, "Longitude": -84.5039 },
    { "Name": "Christ Hospital", "Latitude": 39.1231, "Longitude": -84.507 },
    { "Name": "Cleveland Clinic", "Latitude": 41.5033, "Longitude": -81.62 },
    { "Name": "Mount Carmel East", "Latitude": 39.9828, "Longitude": -82.8291 },
]
```

```python
# 4. Structure user inputs
patient_condition = {
    "Age": age,
    "Mobility Issues": mobility,
    "Known Allergies": [a.strip() for a in allergies.split(",") if a.strip()],
    "Chronic Conditions": [c.strip() for c in conditions.split(",") if c.strip()],
    "Recent Surgeries": surgeries,
    "Immunization Status": immunization,
    "Other Notes": other_notes
}
```

```python

geographic_location = {
    "Country": "United States",
    "State": state,
    "City": city,
    "Preferred Mode of Care": mode_of_care
}
```

```python
symptom_list = [s.strip() for s in symptoms.split(",") if s.strip()]
```

```python
# 5. Construct prompt
prompt = f"""
You are a healthcare AI assistant tasked with generating a customized treatment plan for a patient. Given the following inputs:

1. Symptoms: {json.dumps(symptom_list)}
2. Physical Condition: {json.dumps(patient_condition, indent=2)}
3. Location: {json.dumps(geographic_location, indent=2)}
4. Hospitals: {json.dumps(hospital_data, indent=2)}

Generate a treatment plan that includes:

1. Medical Actions
2. Location-Specific Options
3. Justifications

Be concise, medically accurate, and context-aware. Avoid unnecessary medical jargon.
"""
```

```python
from dotenv import load_dotenv
from google import genai
load_dotenv()
client = genai.Client(api_key= os.getenv('Gemini_Api_Key'))
# 6. Generate content using Gemini
response = client.models.generate_content(
    model="gemini-2.0-flash", contents= prompt
)
print("\n===== AI-Generated Treatment Plan =====\n")
print(response.text)
```

```python
# Display the output in a clean, formatted manner
print("\n===== Customized Treatment Plan =====\n")

# Split sections if the model followed the prompt structure
sections = response.text.strip().split('\n\n')

for section in sections:
    if section.strip():
        print(section)
        print("-" * 60)  # Divider between sections
```
